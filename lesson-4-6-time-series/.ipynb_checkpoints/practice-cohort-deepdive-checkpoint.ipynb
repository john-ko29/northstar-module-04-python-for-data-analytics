```python
# Import libraries needed
import pandas as pd
import numpy as np

# Set options on pd
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

# Load CSV file needed
df = pd.read_csv(
    'finflow-enriched-200k.csv',
    # Change datatype to date
    parse_dates=['signup_date', 'last_login_date', 'churn_date']
)

# Check on file loaded correctly
print(f"Loaded {len(df):,} rows")
print(f"Signup date range: {df['signup_date'].min()} to {df['signup_date'].max()}")
print(f"Columns: {df.shape[1]}")
```

## Question 1: The cohort-by-tier cross-cut.

For each combination of (signup_cohort, marketing_name), compute churn_rate_pct and customer_count. Use a multi-column groupby. Unstack the result so cohorts are rows and tiers are columns. The shape should be 25 rows by 6 columns (3 tiers, 2 metrics each, after unstacking).

Save as cohort-by-tier-breakdown.csv.

```python
# Create signup cohort = month the customer signup
df['signup_cohort'] = df['signup_date'].dt.to_period('M')

# Check on signup_cohort
print(df[['customer_id', 'signup_date', 'signup_cohort']].head())
print()
print(f"Type of signup_cohort: {df['signup_cohort'].dtype}")
print(f"Number of distinct cohorts: {df['signup_cohort'].nunique()}")
```

```python
cohort_tier = (
    df.groupby(['signup_cohort', 'marketing_name']).agg(
        churn_rate_pct=('active', lambda s: (1 - s.mean()) * 100),
        customer_count=('customer_id', 'size')
    ).unstack('marketing_name').round(2)
)

# Check results
print("Cohort tier shape:")
print(cohort_tier.shape)
print("Cohort tier rows:")
# Check the first 5 rows of the data
print(cohort_tier)
```

```python
# Export table to CSV file
cohort_tier.to_csv('cohort-by-tier-breakdown.csv')
print("Saved cohort-by-tier-breakdown.csv")
```

## Question 2: Cohort size trend.
Using resample, produce a Series of total signups per quarter ('QE' frequency in Pandas 2.x, or 'Q' in older versions). Print the result. Did signups grow each quarter, or were some quarters down?

```python
# Resample DataFrame to quarterly 
quarterly_signups = df.set_index('signup_date').resample('QE').size()

# Print Results
print(quarterly_signups)

# Check difference between each quarter
print(quarterly_signups.diff())
```

## Question 3: A six-month rolling cohort churn average.
Take your cohort_churn table from the Guided Example. Compute a six-month rolling average of churn_rate_pct. Add it as a new column called churn_rate_pct_6mo_rolling. The first five rows of the new column will be NaN; that's expected.

This smoothed series is what product leadership would actually want to see as a trend line. The cohort-by-cohort churn rate is noisy because individual months can have idiosyncratic events (a marketing push, a product launch, a holiday). A six-month rolling average shows whether the underlying trend is improving or worsening.

Grab cohort_churn from Guided Example

```python
def churn_rate_pct(active_series):
    return (~active_series).mean() * 100

cohort_churn = df.groupby('signup_cohort').agg(
    customer_count=('customer_id', 'count'),
    churn_rate_pct=('active', churn_rate_pct),
    avg_monthly_fee=('monthly_fee', 'mean')
).round(2)

print(cohort_churn.head(10))
print("--- (rows omitted) ---")
print(cohort_churn.tail(5))
```

```python
# Average of the current cohort and the 5 before it
# the first 5 rows are NaN
cohort_churn['churn_rate_pct_6mo_rolling'] = (
    cohort_churn['churn_rate_pct'].rolling(window=6).mean()
)

print(cohort_churn.round(2))
```
