## Set up and Imports

````python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

df = pd.read_csv(
    'finflow-enriched-200k.csv',
    parse_dates=['signup_date', 'last_login_date', 'churn_date']
)

print(f"Loaded {len(df):,} rows")
print(f"Columns: {list(df.columns)}")
````

## Step 1: Define the calculation

* For each churned customer (`active == False`), **days active** = `(churn_date - signup_date).dt.days`
* Compute the distribution of days-active per plan tier (`marketing_name`): **mean, median, p25, p75, count**
* **Output:** one table, one row per `marketing_name` (expected: 3 rows) and five columns of statistics (expected: 3 x 5)

## Step 2: Prompt AI for the scaffolding.



Prompt: 
I have a Pandas DataFrame df with columns including marketing_name (string), signup_date (datatime), churn_date (datatime), plan_tier_clean (string), and active (boolean).
Compute the following:
1. For each churned customer, "days active" is (churn_date - signup_date).dt.days
2. Compute the distribution of days-active per plan tier (mean, median, p25, p75, count)
3. Output: a table with one row per marketing_name and five columns of statistics
Explain what each step does.

AI response:

```python
churned = df[~df["active"]].copy()
churned["days_active"] = (churned["churn_date"] - churned["signup_date"]).dt.days

summary = (
    churned.groupby("marketing_name")["days_active"]
    .agg(
        mean="mean",
        median="median",
        p25=lambda s: s.quantile(0.25),
        p75=lambda s: s.quantile(0.75),
        count="count",
    )
    .round(1)
)
summary
```

**What each step does**

1. **Filter to churned customers.** `~df["active"]` flips the booleans, so it keeps only rows where `active` is False. `.copy()` makes an independent DataFrame so adding a column in the next line doesn't trigger a `SettingWithCopyWarning`.
2. **Compute days active.** Subtracting two datetime columns gives a timedelta for each row. `.dt.days` converts each timedelta to a whole number of days and stores it in a new `days_active` column.
3. **Group by plan tier.** `groupby("marketing_name")["days_active"]` splits the churned customers into one group per `marketing_name` and selects the `days_active` column.
4. **Aggregate.** `.agg(...)` uses named aggregation, where each keyword becomes an output column. `mean`, `median` and `count` are built-in. Pandas has no built-in percentile string, so p25 and p75 use small lambdas around `Series.quantile`.
5. **Round.** `.round(1)` rounds every statistic to one decimal place for readability.

The result has one row per `marketing_name` and five columns: mean, median, p25, p75, count.

**Things to watch for**

- **`active` must be a real boolean.** `~` only works if `active` loaded as `bool`. If it loaded as text, you'll get an error or wrong results.
- **Dates must be real datetimes.** If `signup_date` or `churn_date` are still strings, the subtraction fails with a `TypeError`. Load them with `parse_dates=[...]` or convert them with `pd.to_datetime`.
- **`count` ignores missing values.** It counts churned customers with a valid `days_active`, not all churned rows.
- **The code doesn't check that `churn_date` is on or after `signup_date`.** If some rows are the other way round, they produce negative days and pull the mean and percentiles down. Check with `(churned["days_active"] < 0).sum()`.
- **You mentioned `plan_tier_clean` too.** You asked for one row per `marketing_name`, so I grouped by that. If you meant the cleaned tier names instead, swap in `groupby("plan_tier_clean")`.

## Step 3 Apply the verification protocol

````python
# Step 1: keep only churned customers
# .copy() avoids SettingWithCopyWarning when adding a column next
churned = df[~df["active"]].copy()

# Step 2: Days active = churn date minus signup date
# .dt.days converts them to whole days.
churned["days_active"] = (churned["churn_date"] - churned["signup_date"]).dt.days

# Step 3: Build the summary table, one row per marketing_name.
summary = (
    churned
    # # Split churned customers into one group per plan tier and focus on days_active
    .groupby("marketing_name")["days_active"]
    # Named aggregation: each keyword becomes an output column
    .agg(
        mean="mean", # average days active
        median="median", # # middle value (50th percentile)
        p25=lambda s: s.quantile(0.25), # 25th percentile (no built-in string, so use a lambda)
        p75=lambda s: s.quantile(0.75), # 75th percentile
        count="count",  # number of churned customers with a valid days_active
    )
    # Round all statistics to 1 decimal place
    .round(1)
)

# Display the result
summary
````

````python
df_small = df.head(500)
churned = df_small[~df_small["active"]].copy()
churned["days_active"] = (churned["churn_date"] - churned["signup_date"]).dt.days

summary = (
    churned.groupby("marketing_name")["days_active"]
    .agg(
        mean="mean",
        median="median",
        p25=lambda s: s.quantile(0.25),
        p75=lambda s: s.quantile(0.75),
        count="count",
    )
    .round(1)
)
print(summary.shape)
summary
````

A `KeyError: 'churn'` means pandas looked for a column labeled exactly `'churn'` and none exists. Pandas matches column names exactly, so even a small difference in spelling or capitalization will trigger it.

**The likely cause:** your DataFrame has a column called `churn_date`, not `churn`. The line that fails probably looks something like this:

```python
churned["days_active"] = (churned["churn"] - churned["signup_date"]).dt.days
```

**The fix:** use the exact column name.

```python
churned["days_active"] = (churned["churn_date"] - churned["signup_date"]).dt.days
```

**If it's not that line,** you can find the exact names with:

```python
print(df.columns.tolist())
```

Copy the name from that list, since trailing spaces or different capitalization (`Churn_Date`) cause the same error.

**Reading the traceback:** the long block of lines above the last one is just pandas' internal lookup. The last line, `KeyError: 'churn'`, tells you the label that wasn't found. The line number near the top of the traceback points to where in your code the lookup failed.
