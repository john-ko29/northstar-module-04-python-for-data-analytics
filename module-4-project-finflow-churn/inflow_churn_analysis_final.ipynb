## Section 3
Per-tier churn breakdown

```python
# Custom function: churn rate as a percentage
def churn_rate_pct(active_series):
    return (~active_series).mean() * 100

# Apply via agg() alongside other metrics
tier_churn_summary = df.groupby('marketing_name').agg({
    'customer_id': 'count',
    'monthly_fee': 'sum',
    'active': churn_rate_pct
})
# Set column names
tier_churn_summary.columns = ['customer_count', 'total_mmr', 'churn_rate_pct']
# Check results
print("Per-tier summary:")
print(tier_churn_summary)
```

## Section 4
Signup-cohort breakdown

```python
# create a column with each signup date to a monthly period
df['signup_cohort'] = df['signup_date'].dt.to_period('M')

# Create a Dataframe using agg function
cohort_churn = df.groupby('signup_cohort').agg(
    customer_count=('customer_id', 'count'),
    churn_rate_pct=('active', churn_rate_pct)
).round(2)

# Display DataFrame
cohort_churn
```

All the months have similiar churn rate around 20%. There is not definite pattern that I can discern from just the churn rate to signup_cohort alone. 

## Section 5: 
Cross-cut, cohort by tier.

```python
# Group rows by signup month AND plan tier together then select the churn_date column inside each group.
cohorot_by_tier = df.groupby(['signup_cohort', "plan_tier"])["churn_date"]

# Counts only NON-null churn_date values in each group
churned_n = cohorot_by_tier.count().unstack("plan_tier")
# Counts total amount of customer 
customers_n = cohorot_by_tier.size().unstack("plan_tier")

# Find churn percentage 
cohort_by_tier = (churned_n / customers_n * 100).round(2)
# Remove 'plan_tier' so it removes the extra header row
cohort_by_tier.columns.name = None
# Display the table
cohort_by_tier
```
