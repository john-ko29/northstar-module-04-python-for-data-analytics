## Section 1

```python
# Import libraries needed 
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set options for readability 
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

# Load CSV file to DataFrame
df = pd.read_csv(
    'finflow-enriched-200k.csv',
    # Change datatype to datetime
    parse_dates=['signup_date', 'last_login_date', 'churn_date']
)
```

## Section 3
Per-tier churn breakdown

```python
# Custom function: churn rate as a percentage
def churn_rate_pct(active_series):
    return (~active_series).mean() * 100

# Apply via agg() alongside other metrics
tier_churn_summary = df.groupby('marketing_name').agg({
    'customer_id': 'count',
    'monthly_fee': 'sum',
    'active': churn_rate_pct
})
# Set column names
tier_churn_summary.columns = ['customer_count', 'total_mmr', 'churn_rate_pct']
# Check results
print("Per-tier summary:")
print(tier_churn_summary)
```

## Section 4
Signup-cohort breakdown

```python
# create a column with each signup date to a monthly period
df['signup_cohort'] = df['signup_date'].dt.to_period('M')

# Create a Dataframe using agg function
cohort_churn = df.groupby('signup_cohort').agg(
    customer_count=('customer_id', 'count'),
    churn_rate_pct=('active', churn_rate_pct)
).round(2)

# Display DataFrame
cohort_churn
```

All the months have similiar churn rate around 20%. There is not definite pattern that I can discern from just the churn rate to signup_cohort alone. 

## Section 5: 
Cross-cut, cohort by tier.

```python
# Group rows by signup month AND plan tier together then select the churn_date column inside each group.
cohorot_by_tier = df.groupby(['signup_cohort', "plan_tier"])["churn_date"]

# Counts only NON-null churn_date values in each group
churned_n = cohorot_by_tier.count().unstack("plan_tier")
# Counts total amount of customer 
customers_n = cohorot_by_tier.size().unstack("plan_tier")

# Find churn percentage 
cohort_by_tier = (churned_n / customers_n * 100).round(2)
# Remove 'plan_tier' so it removes the extra header row
cohort_by_tier.columns.name = None
# Display the table
cohort_by_tier
```

## Section 6: Summary of Headline Findings
Overall churn is 19.98%, 39,951 of the 200,000 customers have churned.
Churn is almost identical across plan tiers. Starter Plan (Basic) is 19.93%, Professional (Pro) is 20.09%, and Enterprise Plus (Enterprise) is 19.88%. 
No signup cohort has elevated churn. The 24 full monthly cohorts (January 2024 to December 2025) range from 19.23% (February 2025) to 20.61% (June 2025), all close to the 20% overall rate. The January 2026 cohort has only 287 customers and is too small to compare.
he cohort-by-tier table shows no tier-specific problem months. Starter Plan and Professional stay near 20% in every cohort. Enterprise Plus ranges from about 17% to 23%, but each monthly cohort has only about 700 to 900 customers, so those swings are mostly small-sample noise.

```python
# Panel 1: bar, churn rate per tier
deck.add_trace(go.Bar(                                # add a bar chart to the grid
    x=tiers, y=tier_rate.values,                      # plans on x, churn rate on y
    showlegend=False,
), row=1, col=1)                                      # place it in row 1, column 
deck.update_xaxes(title_text='Tier', row=1, col=1)
deck.update_yaxes(title_text='Churn rate', row=1, col=1)

# Panel 2: donut, customer mix
deck.add_trace(go.Pie(                                # add a pie chart
    labels=tiers, values=tier_n.values,     # slices are plans; hole=0.5 makes it a donut
), row=1, col=2)                                      # place it in row 1, column 2

# Panel 3: line, cohort churn rate (full cohorts only)  -> row 2, col 1
deck.add_trace(go.Scatter(                            # add a line chart
    x=cohort_full.index.to_timestamp(), y=cohort_full["rate"],  # signup months on x (as dates), churn rate on y
    mode="lines+markers", line=dict(color="firebrick"), showlegend=False,  # line with dots, red, no legend
), row=2, col=1)                                      # place it in row 2, column 1
deck.update_xaxes(title_text='Signup month', row=2, col=1)
deck.update_yaxes(title_text='Churn rate (%)', row=2, col=1)

# Panel 4: heatmap, cohort x tier  -> row 2, col 2
deck.add_trace(go.Heatmap(                            # add a heatmap
    z=cohort_tier.values, x=tiers, y=cohort_tier.index.astype(str),  # cell values, plan columns, month rows
    colorscale="Reds", zmin=0, zmax=30,               # white to red, fixed 0-30% so "flat" looks flat
    colorbar=dict(title="%", thickness=8, len=0.25, x=1.02, y=0.5),  # small color legend at the right edge, beside this row
), row=2, col=2)                                      # place it in row 2, column 2
deck.update_xaxes(title_text="Plan", row=2, col=2)    # label the x-axis
deck.update_yaxes(title_text="Signup month", autorange="reversed", row=2, col=2)  # label y-axis, earliest month at the top

# Panel 5: two lines, cumulative active vs churned  -> row 3, col 1
x5 = cum.index.to_timestamp()                         # convert month labels to dates for the x-axis
for name, color in [("Active", "steelblue"), ("Churned", "firebrick")]:  # draw one line per status
    deck.add_trace(go.Scatter(                        # add a line chart trace
        x=x5, y=cum[name], mode="lines+text", name=name,  # cumulative values, line plus a text label
        line=dict(color=color, width=3),              # thick line in the status color
        showlegend=False,    # label above-left of the point, no legend
    ), row=3, col=1)                                  # place it in row 3, column 1
deck.update_xaxes(title_text="Signup month", tickformat="%b %y", dtick="M3", row=3, col=1)  # label x, ticks every 3 months
deck.update_yaxes(title_text="Cumulative customers", row=3, col=1)  # label the y-axis

# Set deck layout setting
deck.update_layout(
    title_text='FinFlow Customer Analytics: Cohort and Tier Patterns',
    height=1000,
    showlegend=False
)

# Panel 6: churn rate by support tickets  -> row 3, col 2
deck.add_trace(go.Bar(                                # add a bar chart
    x=tickets.index, y=tickets["rate"],               # ticket counts on x, churn rate on y
    customdata=tickets["n"],                          # attach customer counts for the tooltip
    showlegend=False,                                 # no legend needed
), row=3, col=2)                                      # place it in row 3, column 2
deck.update_xaxes(title_text="Support tickets opened", dtick=1, row=3, col=2)  # label x, show every whole number
deck.update_yaxes(title_text="Churn rate (%)", range=[0, 120], row=3, col=2)   # label y, headroom above 100% for bar labels
```

```python
# Export charts to an HTML file and print to confirm the process
deck.write_html("finflow_churn_dashboard.html")
print("finflow_churn_dashboard.html saved")
```
