1. Setup
Import tools needed and define the boundarys of our synthetic dataset

```python
# Imports
from faker import Faker, VERSION
import random
import csv
from datetime import date, timedelta
from pathlib import Path
import time

# Set seeds for reproducibility.
# The number 42 is conventional. Use the same value every time you want
# the same output. Change it if you want a different "random" run.
SEED = 42
Faker.seed(SEED)
random.seed(SEED)

# Pin the dataset to a fixed "as of" date.
AS_OF_DATE = date(2026, 1, 1)

# Customers sign up within the 2 years before the as-of date.
SIGNUP_WINDOW_DAYS = 730
SIGNUP_WINDOW_START = AS_OF_DATE - timedelta(days=SIGNUP_WINDOW_DAYS)

# Set plan fee per tier 
PLAN_FEES = {'Basic': 9.99, 'Pro': 29.99, 'Enterprise': 199.00}

#Extension 3
# Churn probability by plan tier
CHURN_RATES = {'Basic': 0.25, 'Pro': 0.15, 'Enterprise': 0.05}

# Initialize the Faker instance with US English locale.
fake = Faker('en_US')

print("Faker initialized with seed", SEED, "| Faker version", VERSION)
print("Test draw:", fake.name())
```

## Create a method to return a customer with populated fields

```python
def build_one_customer_v3(index):
    # Pick a random first and last name
    first = fake.first_name()
    last = fake.last_name()

    # Extension 1
    # Assign a sign-up date using a weighted date 
    days_ago = int(random.triangular(0, SIGNUP_WINDOW_DAYS, 180))
    signup_date = AS_OF_DATE - timedelta(days=days_ago)

    #Pick a plan: 60% Basic, 30% Pro, 10% Enterprise
    plan_tier = random.choices(
        ['Basic', 'Pro', 'Enterprise'], weights=[60, 30, 10], k=1
    )[0]
    # Find monthly fee based on plan tier assigned
    monthly_fee = PLAN_FEES[plan_tier]

    # Extension 2 
    # Pick referral source 
    referral_source = random.choices(
        ['Organic Search', 'Paid Search', 'Social Media', 'Word of Mouth', 'Affiliate', 'Direct'],
        weights=[45, 20, 15, 10, 7, 3], k=1
    )[0]

    # Extension 3
    # Churn rate based on plan tier
    churn_rate = CHURN_RATES[plan_tier]
    is_active = random.choices(
        [True, False], weights=[1-churn_rate, churn_rate], k=1
    )[0]

    # Assign today as AS_OF_DATE
    today = AS_OF_DATE

    # last_login_date and churn_date depend on active state,
    # measured from the pinned as-of date rather than the clock
    if is_active:
        # Active customers logged in within the last 30 days
        days_since_login = random.randint(0, 30)
        last_login = today - timedelta(days=days_since_login)
        churn_date = None
    else:
        # Churned customers haven't logged in for 30 to 180 days
        days_since_login = random.randint(30, 180)
        last_login = today - timedelta(days=days_since_login)
        # Churn date is somewhere between last_login and today
        days_since_churn = random.randint(0, days_since_login)
        churn_date = today - timedelta(days=days_since_churn)

    # Engagement metrics: correlate with active state
    if is_active:
        total_sessions = random.randint(20, 400)
        avg_session_duration = round(random.uniform(3.0, 15.0), 1)
        support_tickets = random.randint(0, 4)
    else:
        total_sessions = random.randint(1, 50)
        avg_session_duration = round(random.uniform(1.0, 6.0), 1)
        support_tickets = random.randint(0, 8)

    return {
        "customer_id": f"FF-{index:05d}",
        "first_name": first,
        "last_name": last,
        "email": f"{first.lower()}.{last.lower()}{index}@example.com",
        "city": fake.city(),
        "state": fake.state_abbr(),
        "country": "US",
        "signup_date": signup_date.isoformat(),
        "plan_tier": plan_tier,
        "monthly_fee": monthly_fee,
        "referral_source": referral_source,
        "last_login_date": last_login.isoformat(),
        "total_sessions": total_sessions,
        "avg_session_duration_minutes": avg_session_duration,
        "support_tickets_opened": support_tickets,
        "active": is_active,
        "churn_date": churn_date.isoformat() if churn_date else None,
    }


# Build and display a single record
sample = build_one_customer_v3(1)
for field, value in sample.items():
    print(f"  {field}: {value}")
```

## Generate 1,000 customers
Generate 1,000 customers

```python
# Re-seed so the run is reproducible from this point
Faker.seed(SEED)
random.seed(SEED)

# Generate 1,000 customers
start = time.time()
customers = [build_one_customer_v3(i) for i in range(1, 1001)]
elapsed = time.time() - start

print(f"Built {len(customers)} customers in {elapsed:.2f} seconds")
print()
print("First customer:")
for field, value in customers[0].items():
    print(f"  {field}: {value}")
print()
print("Last customer:")
for field, value in customers[-1].items():
    print(f"  {field}: {value}")
```

## Sanity checks

Before saving, verify that the generated data actually matches the
distributions we asked for. Plan tier weights, referral sources, sign-up dates, churn rate, value ranges.

```python
#Helper 
def to_date(text):
    return date.fromisoformat(text)

# Check 1: Plan tier distribution
plan_counts = {}
for c in customers:
    tier = c["plan_tier"]
    plan_counts[tier] = plan_counts.get(tier, 0) + 1

print("Plan tier distribution (expected: ~60/30/10):")
for tier in ['Basic', 'Pro', 'Enterprise']:
    count = plan_counts.get(tier, 0)
    pct = count / len(customers) * 100
    print(f"  {tier}: {count} ({pct:.1f}%)")

# Check 2: Referral Source distribution
referral_counts = {}
for c in customers:
    referral = c["referral_source"]
    referral_counts[referral] = referral_counts.get(referral, 0) + 1

print(f"\nReferral source distribution (expected: ~45/20/15/10/7/3):")
for referral in ['Organic Search', 'Paid Search', 'Social Media', 'Word of Mouth', 'Affiliate', 'Direct']:
    count = referral_counts.get(referral, 0)
    pct = count / len(customers) * 100
    print(f"  {referral}: {count} ({pct:.1f}%)")

# Check 3: Signup recency distribution (targets: ~60% in last 12 months, ~90% in last 18)
days_ago_list = [(AS_OF_DATE - to_date(c['signup_date'])).days for c in customers]
last_12 = sum(1 for d in days_ago_list if d <= 365)
months_12_24 = sum(1 for d in days_ago_list if d > 365)
last_18 = sum(1 for d in days_ago_list if d <= 548)
print(f"\nSignup dates last 12 months: ~60% last 18 months: ~90%")
print(f"Signed up in last 12 months:   {last_12} ({last_12 / len(customers):.1%})")
print(f"Signed up 12-24 months ago:    {months_12_24} ({months_12_24 / len(customers):.1%})")
print(f"Signed up in last 18 months:   {last_18} ({last_18 / len(customers):.1%})")

# Check 4: Churn rate by plan tier
churn_rate_by_tier = {}
print(f"\nChurn rate by plan tier (expected: ~25//15/5)")
for tier in ['Basic', 'Pro', 'Enterprise']:
    in_tier = [c for c in customers if c['plan_tier'] == tier]
    churned_count = sum(1 for c in in_tier if not c['active'])
    churn_rate_by_tier[tier] = churned_count / len(in_tier)
    print(f"  {tier}: {churn_rate_by_tier[tier]:.1%} ({churned_count} of {len(in_tier)})")

# Monthly fee total
total_mrr = sum(c["monthly_fee"] for c in customers if c["active"])
print(f"\nMonthly recurring revenue from active customers: ${total_mrr:,.2f}")
```

**Observation:** Distributions look reasonable. 
Plan tier proportions are within 2 percentage point of the 60/30/10 target. 
Referral source proportions are about 1.5 percentage point of the 45/20/15/10/7/3 target
Signup dates are about at most 7.5 point away from 60% while 18 months is about 2.4 percentage point away from 90% target
Churn rate by plan tier is about 1 percentage point away from 25//15/5

## Saving to CSV

Save the generated customers as a CSV file in the lesson folder.

```python
# Define the output path. Path() handles Mac/Windows path differences cleanly.
output_path = Path("finflow-customers-1000-extended.csv")

# Get the field names from the first record (Python 3.7+ dicts preserve order)
field_names = list(customers[0].keys())

# Write the CSV
with open(output_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=field_names)
    writer.writeheader()
    writer.writerows(customers)

# Verify
print(f"Wrote {len(customers)} customers to {output_path.resolve()}")
print(f"File size: {output_path.stat().st_size:,} bytes")
```
