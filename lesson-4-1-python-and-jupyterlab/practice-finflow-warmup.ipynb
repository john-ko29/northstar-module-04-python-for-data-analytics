## Analysis 1

Count the total number of active customers and the total number of churned (active == False) customers. Print both. Add a markdown observation noting the churn rate as a percentage.

```python
# Count active vs. churned with a loop
active_count = 0
churned_count = 0
for fs in finflow_sample:
    if fs["active"]:
        active_count = active_count + 1
    else:
        churned_count = churned_count +1

print(f"Active customers: {active_count}" )
print(f"Churned customers: {churned_count}")
```

The number of active customers is 6 and number of churned customers is 4
Active customer: 60%
Churned customers: 40%

## Analysis 2

Compute the total monthly revenue from currently active customers. Print it formatted as US dollars. Add a markdown observation noting how Enterprise customers compare to Basic in terms of contribution.

```python
# Add all 
active_sum_revenue = sum(fs["monthly_fee"] for fs in finflow_sample if fs["active"])
print(f"Total monthly revenue from active customers: ${active_sum_revenue}")
```

From the total monthly revenue from active customers:

## Analysis 3

For each plan tier (Basic, Pro, Enterprise), print the count of churned customers in that tier. Use a loop or a dictionary. Add a markdown observation about which tier appears most vulnerable to churn.

```python
# Count the number of churned customers in each plan tier with a loop
basic_churn = 0
pro_churn = 0
enterprise_churn = 0
for fs in finflow_sample:
    if not fs["active"]:
        if fs["plan_tier"] == "Basic":
            basic_churn = basic_churn + 1
        elif fs["plan_tier"] == "Pro":
            pro_churn = pro_churn + 1
        elif fs["plan_tier"] == "Enterprise":
            enterprise_churn = enterprise_churn + 1
            

print(f"Basic churned: {basic_churn}")
print(f"Pro churned: {pro_churn}")
print(f"Enterprise churned: {enterprise_churn}")
```

There are sample size of 4 so it's not reliable but basic has the most at 2. Basic appears the most vulnerable to churn. With low months active, they are quick to churn.

## Analysis 4

 Compute the average months_active for churned customers. Print it. Add a markdown observation about what this number suggests (do customers churn early, mid-tenure, or late?).

```python
# Add months_active if false for active
churned_months = [fs["months_active"] for fs in finflow_sample if not fs["active"]]
avg_churned_months = sum(churned_months) / len(churned_months)

print(f"Average month active before churned: {avg_churned_months}")
```

The number suggests that most customers churn early. The lower the tier the more, the earlier the customer is most likely to churn.

## Analysis 5

Write a function named lifetime_revenue(customer) that takes a customer dictionary and returns monthly_fee * months_active as the customer's lifetime revenue so far. Use the function to compute the total lifetime revenue across all ten customers. Print the total formatted as US dollars. Add a markdown observation about what this represents.

```python
# Return lifetime_revenue of a customer 
def lifetime_revenue(customer):
    return customer["monthly_fee"] * customer["months_active"]

# Enter each store and print out the result 
for fs in finflow_sample:
    total_revenue = lifetime_revenue(fs)
    print(f"{fs["customer_id"]} has a total lifetime revenue of ${total_revenue:.2f}")
```

This data represents how much FinFlow has collected from each customer in the data.
